Read course_catalog.json. Turn courses into useful documents while retaining IDs, fees, durations, and other fields.

## Split into chunks
Choose chunk size and overlap. Print representative chunks and verify that source metadata survives.


## Build the vector index
Choose an embedding provider and vector store. Create your index and save it under outputs/.


## Retrieve evidence
Ask a practice question. Print the retrieved chunks, scores if available, and source filenames.


## Generate grounded answers
Add an LLM. Answer from retrieved evidence, cite source filenames, and handle missing information.


## Evaluate and experiment
Read evaluation/questions.json only for evaluation. Compare results and try different chunk sizes and retrieval counts.
