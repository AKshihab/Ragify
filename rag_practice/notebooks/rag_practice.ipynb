```python
from langchain_openai import OpenAIEmbeddings
from langchain_community.vectorstores import FAISS

embeddings = OpenAIEmbeddings(
    api_key=euri_api_key,
    base_url="https://api.euron.one/api/v1/euri",
    model="text-embedding-3-small",
    check_embedding_ctx_length=False,
    encoding_format="float",
)

# Test your key and model access.
vector = embeddings.embed_query("Hello, I am Shihab.")
print("Embedding dimensions:", len(vector))

# Embed your chunks and build the vector store.
vectorstore = FAISS.from_documents(chunks_of_text, embeddings)
```

```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnablePassthrough

llm = ChatOpenAI(
    api_key=euri_api_key,
    base_url="https://api.euron.one/api/v1/euri",
    model="gpt-4.1-mini",
    temperature=0,
)

template = """Answer the question using only the following context.
If the context does not provide the answer, say:
"I don't know based on these documents."

Context:
{context}

Question: {question}
"""

prompt = ChatPromptTemplate.from_template(template)

def format_docs(docs):
    return "\n\n".join(doc.page_content for doc in docs)

rag_chain = (
    {
        "context": retriever | format_docs,
        "question": RunnablePassthrough(),
    }
    | prompt
    | llm
    | StrOutputParser()
)

answer = rag_chain.invoke("Which Agentic AI tools does Shihab use?")
print(answer)
```
